# Bayesian Neural Network Regression
Using Housing_Prices.csv dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
import tensorflow as tf
import tensorflow_probability as tfp
from tensorflow.keras import layers

# Load dataset
url = "https://raw.githubusercontent.com/Wimran681/AdvanceMLearning/main/Assignment01/Housing_Prices.csv"
data = pd.read_csv(url)
data.head()
```

## Preprocessing

```python
# Drop rows with missing target (Price)
data = data.dropna(subset=["Price"])

# Fill or drop missing values
data = data.dropna()  # Simple approach for now

# Encode categorical columns
data = pd.get_dummies(data)

# Separate features and target
X = data.drop("Price", axis=1)
y = data["Price"]

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

## Define Bayesian Neural Network

```python
tfd = tfp.distributions
tfpl = tfp.layers

model = tf.keras.Sequential([
    tfpl.DenseFlipout(64, activation='relu'),
    tfpl.DenseFlipout(64, activation='relu'),
    tfpl.DenseFlipout(1)
])

model.compile(optimizer='adam', loss='mse')
model.summary()
```

## Train Model

```python
history = model.fit(X_train, y_train, epochs=100, batch_size=32, validation_split=0.1)
```

## Evaluate

```python
y_pred = model(X_test).numpy().flatten()
mse = mean_squared_error(y_test, y_pred)
print("Test MSE:", mse)
```
